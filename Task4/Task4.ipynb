## Задание № 1

```python
# Импортируем модуль для работы с регулярными выражениями
import re

# Запишем регулярное выражение для проверки формата номера
# Допустимые буквы: А, В, Е, К, М, Н, О, Р, С, Т, У, Х (кириллица), \d{3} - любые три цифры,
# \d{2,3} - от 2 до 3 цифр
def check_autonumber(car_id):
    pattern = r'^([АВЕКМНОРСТУХ])(\d{3})([АВЕКМНОРСТУХ]{2})(\d{2,3})$' # у нас 4 группы - буквенная комбинация, цифры, буквы, цифры
    match = re.match(pattern, car_id) # проверка номера на соответствие
    
    if match:
        # Извлекаем группы: комбинации буква, цифры, буквы, регион (цифры)
        letter = match.group(1)
        digits = match.group(2)
        letters = match.group(3)
        region = match.group(4)
        
        # Формируем номер
        number = f"{letter}{digits}{letters}"
        
        return f"Номер {number} валиден. Регион: {region}."
    else:
        return "Номер не валиден."
```

```python
car_id = 'А222ВС96'
check_autonumber(car_id)
```

```python
car_id = 'АБ22ВВ193'
check_autonumber(car_id)
```

## Задание № 2

```python
import re

def remove_duplicate(text):
    # Регулярное выражение для поиска последовательных повторяющихся слов
    # \b - граница слова, (\w+) - слово (1+ букв), \s+ - пробелы, \1 - повтор первой группы
    pattern = r'\b(\w+)\s+\1\b'
    
    # В цикле заменяем повторы на одно слово, пока есть повторы
    while re.search(pattern, text):
        text = re.sub(pattern, r'\1', text)
    
    return text

# Пример использования
string = 'Напишите функцию функцию, которая будет будет будет будет удалять все все все все последовательные повторы слов из из из из заданной строки строки при помощи регулярных выражений'

remove_duplicate(string)
```
